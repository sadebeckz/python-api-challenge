## Starter Code to Import Libraries and Load the Weather and Coordinates Data

```python
# Dependencies and Setup
import hvplot.pandas
import pandas as pd
import requests

# Import API key
from api_keys import geoapify_key
```

```python
# Load the CSV file created in Part 1 into a Pandas DataFrame
city_data_df = pd.read_csv("output_data/cities.csv")

# Display sample data
city_data_df.head()
```

---

### Step 1: Create a map that displays a point for every city in the `city_data_df` DataFrame. The size of the point should be the humidity in each city.

```python
%%capture --no-display
import hvplot.pandas

# Configure the map plot using hvPlot only
city_map = city_data_df.hvplot.points(
    'Lng',                                
    'Lat',                                
    geo=True,                        
    size='Humidity',                      
    scale=1,                              
    color='Humidity',                     
    tiles='EsriImagery',                  
    frame_width=800,                     
    frame_height=500,                     
    alpha=0.7,                            
    hover_cols=['City', 'Country', 'Humidity']  
)

# Display the map
city_map
```

### Step 2: Narrow down the `city_data_df` DataFrame to find your ideal weather condition

```python
# Narrow down cities that fit the ideal weather criteria
ideal_weather_df = city_data_df[
    (city_data_df["Max Temp"] >= 20) &  
    (city_data_df["Max Temp"] <= 30) &  
    (city_data_df["Humidity"] < 70) &   
    (city_data_df["Wind Speed"] < 5) &  
    (city_data_df["Cloudiness"] < 50)   
]

# Drop any rows with null values
ideal_weather_df = ideal_weather_df.dropna()

# Display sample data
ideal_weather_df.head()
```

### Step 3: Create a new DataFrame called `hotel_df`.

```python
# Step 1: Create a new DataFrame called hotel_df with specific columns
hotel_df = ideal_weather_df[["City", "Country", "Lat", "Lng", "Humidity"]].copy()

# Step 2: Add an empty column "Hotel Name" to store hotel information later
hotel_df["Hotel Name"] = ""

# Step 3: Display sample data
hotel_df.head()
```

### Step 4: For each city, use the Geoapify API to find the first hotel located within 10,000 metres of your coordinates.

```python
import requests
import time

# Step 1: Set search radius for hotels (10,000 meters)
radius = 10000  

# Step 2: Define the API parameters
params = {
    "categories": "accommodation.hotel",  
    "limit": 1,                           
    "apiKey": geoapify_key               
}

# Step 3: Print a starting message
print("Starting hotel search")

# Step 4: Iterate through the hotel_df DataFrame to search for hotels
for index, row in hotel_df.iterrows():
    # Get latitude and longitude for each city
    lat = row["Lat"]
    lng = row["Lng"]

    # Update the params with location-based filters
    params["filter"] = f"circle:{lng},{lat},{radius}"  
    
    params["bias"] = f"proximity:{lng},{lat}"          

    # Define the base URL for the Geoapify API
    base_url = "https://api.geoapify.com/v2/places"

    # Step 5: Make the API request
    response = requests.get(base_url, params=params)

    # Step 6: Convert the API response to JSON format
    name_address = response.json()

    # Step 7: Extract the hotel name from the response and store it in the DataFrame
    try:
        hotel_name = name_address["features"][0]["properties"]["name"]
        hotel_df.loc[index, "Hotel Name"] = hotel_name
    except (KeyError, IndexError):
        # If no hotel is found, assign a default message
        hotel_df.loc[index, "Hotel Name"] = "No hotel found"

    # Step 8: Print the result for each city
    print(f"{hotel_df.loc[index, 'City']} - nearest hotel: {hotel_df.loc[index, 'Hotel Name']}")

    # Optional: Pause to respect API rate limits
    time.sleep(1)  # Pause for 1 second between API calls

# Step 9: Display the updated DataFrame with hotel names
hotel_df.head()
```

### Step 5: Add the hotel name and the country as additional information in the hover message for each city in the map.

```python
%%capture --no-display

import hvplot.pandas

# Step 1: Configure the map plot with hotel name and country in the hover info
hotel_map = hotel_df.hvplot.points(
    'Lng',                                
    'Lat',                                
    geo=True,                             
    size='Humidity',                      
    scale=1,                              
    color='Humidity',                     
    tiles='OSM',                          
    frame_width=800,                      
    frame_height=500,                     
    alpha=0.7,                            
    hover_cols=["City", "Country", "Hotel Name", "Humidity"]  
)

# Step 2: Display the map
hotel_map
```
